```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/Week1_GA_dataset.csv")
```

```python
total_unknown = (df == "?").sum().sum()
print(f"Number of values:{total_unknown}")
```

```python
df.replace("?", np.nan, inplace=True)
```
